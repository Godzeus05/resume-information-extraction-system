```python
import gradio as gr
import json
import sys

sys.path.insert(0, "/content")

from app.extractor import extract_resume


def process_resume(file_path):
    if file_path is None:
        return {"error": "Please upload a PDF or DOCX resume."}

    try:
        result = extract_resume(file_path)
        return result

    except Exception as e:
        return {"error": str(e)}


demo = gr.Interface(
    fn=process_resume,
    inputs=gr.File(
        label="Upload Resume",
        file_types=[".pdf", ".docx"],
        type="filepath"
    ),
    outputs=gr.JSON(label="Extracted Information"),
    title="Resume Information Extraction System",
    description=(
        "Upload a PDF or DOCX resume to extract structured information "
        "using a rule-based Python parser. "
        "No external LLM or Generative AI API is used."
    ),
)

demo.launch(share=True)
```
